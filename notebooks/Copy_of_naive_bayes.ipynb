```python
import pandas as pd
```

```python
df1 = pd.read_excel('train.xlsx')
df1.head()
```

```python
df1_test = pd.read_excel('test.xlsx')
df1_test.head()
```

```python

# Define a function to map values
def map_values(value):
    if value in [1, 2]:
        return "negative"
    elif value == 3:
        return "neutral"
    elif value in [4, 5]:
        return "positive"
    else:
        return value

# Apply the function to the column
df1['new_rating'] = df1['rating'].apply(map_values)
df1_test['new_rating'] = df1_test['rating'].apply(map_values)
```

```python
df1.head()
```

```python
df1_test.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.feature_extraction.text import TfidfVectorizer

X_train = df1['title_hindi']
y_train = df1['new_rating']

X_test = df1_test['title_hindi']
y_test = df1_test['new_rating']
# Convert text data into numerical features using TfidfVectorizer for unigram
vectorizer_unigram = TfidfVectorizer()
X_train_unigram = vectorizer_unigram.fit_transform(X_train)
X_test_unigram = vectorizer_unigram.transform(X_test)

# Train Multinomial Naive Bayes model on unigram features
nb_unigram = MultinomialNB()
nb_unigram.fit(X_train_unigram, y_train)

# Predict on test set
y_pred_unigram = nb_unigram.predict(X_test_unigram)

# Calculate accuracy for unigram
accuracy_unigram = accuracy_score(y_test, y_pred_unigram)
print("Accuracy for unigram using TFidf:", accuracy_unigram)
print(classification_report(y_test, y_pred_unigram))

# Repeat the same process for bigram
# Convert text data into numerical features using TfidfVectorizer for bigram
vectorizer_bigram = TfidfVectorizer(ngram_range=(2, 2))
X_train_bigram = vectorizer_bigram.fit_transform(X_train)
X_test_bigram = vectorizer_bigram.transform(X_test)

# Train Multinomial Naive Bayes model on bigram features
nb_bigram = MultinomialNB()
nb_bigram.fit(X_train_bigram, y_train)

# Predict on test set
y_pred_bigram = nb_bigram.predict(X_test_bigram)

# Calculate accuracy for bigram
accuracy_bigram = accuracy_score(y_test, y_pred_bigram)
print("Accuracy for bigram using TFidf:", accuracy_bigram)
print(classification_report(y_test, y_pred_bigram))
```
